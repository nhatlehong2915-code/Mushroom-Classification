```python
# ======================================================================
# CELL 1
# ======================================================================
import os, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

warnings.filterwarnings('ignore')        # tránh in hàng trăm KB cảnh báo làm chậm/lệch số đo

try:
    display                               # Colab/Jupyter
except NameError:
    display = print

XLSX = 'Mushroom_Preprocessed_Dataset_Cleaned.xlsx'   # file đã tiền xử lý (nếu có)
CSV  = 'mushrooms.csv'                                # file gốc UCI (dự phòng)

MINSUP_LIST  = [0.3, 0.4, 0.5]
MINCONF_LIST = [0.6, 0.7, 0.8]
N_REPEAT     = 3          # lặp mỗi cấu hình, lấy median

# --- tham số lọc luật ---
SEL_MINSUP, SEL_MINCONF = 0.3, 0.7     # bộ ngưỡng dùng để chọn luật tiêu biểu
LIFT_LOW, LIFT_HIGH     = 0.9, 1.1     # lift trong khoảng này = "≈ 1" -> loại
N_CLASS_PER_SIDE        = 5            # luật cho mỗi lớp (edible / poisonous)
N_OTHER                 = 6            # luật khác (vế phải không phải class)


# %% ===================== 1. ĐỌC & TIỀN XỬ LÝ =====================
def load_data():
    if os.path.exists(XLSX):
        # PHẢI đọc đúng sheet dữ liệu sạch (sheet 0 chỉ là bảng mô tả 5 dòng)
        d = pd.read_excel(XLSX, sheet_name='3. Dữ liệu đã xử lý (Cleaned)')
    else:
        d = pd.read_csv(CSV)
        d = d.drop(columns=['veil-type'], errors='ignore').drop_duplicates()
    return d

df = load_data()
print(f'-> Dữ liệu: {df.shape[0]} dòng, {df.shape[1]} cột')
df_encoded = pd.get_dummies(df) > 0
print(f'-> Ma trận one-hot: {df_encoded.shape}')


# %% ===================== 2. CHẠY 2 THUẬT TOÁN =====================
def run_once(algo, minsup, minconf):
    t0 = time.time()
    fi = algo(df_encoded, min_support=minsup, use_colnames=True)
    t_mining = time.time() - t0
    rules = association_rules(fi, metric='confidence', min_threshold=minconf)
    t_total = time.time() - t0
    return fi, rules, t_mining, t_total

# warm-up: loại chi phí khởi tạo lần đầu khỏi số đo
apriori(df_encoded, min_support=0.5, use_colnames=True)
fpgrowth(df_encoded, min_support=0.5, use_colnames=True)

rows_ap, rows_fp, rows_cmp = [], [], []
for minsup in MINSUP_LIST:
    for minconf in MINCONF_LIST:
        res = {}
        for name, algo in [('Apriori', apriori), ('FP-Growth', fpgrowth)]:
            tm_l, tt_l = [], []
            for _ in range(N_REPEAT):
                fi, rules, tm, tt = run_once(algo, minsup, minconf)
                tm_l.append(tm); tt_l.append(tt)
            res[name] = dict(fi=fi, rules=rules,
                             t_mining=float(np.median(tm_l)),
                             t_total=float(np.median(tt_l)))
        ap, fp = res['Apriori'], res['FP-Growth']

        rows_ap.append({'Minsup': minsup, 'Minconf': minconf,
                        'Số tập phổ biến': len(ap['fi']), 'Tổng số luật': len(ap['rules']),
                        'Thời gian (giây)': round(ap['t_total'], 4)})
        rows_fp.append({'Minsup': minsup, 'Minconf': minconf,
                        'Số tập phổ biến': len(fp['fi']), 'Tổng số luật': len(fp['rules']),
                        'Thời gian (giây)': round(fp['t_total'], 4)})

        same_fi = set(ap['fi']['itemsets']) == set(fp['fi']['itemsets'])
        same_rules = (set(zip(ap['rules']['antecedents'], ap['rules']['consequents'])) ==
                      set(zip(fp['rules']['antecedents'], fp['rules']['consequents'])))
        speedup = ap['t_total'] / fp['t_total'] if fp['t_total'] > 0 else np.nan
        rows_cmp.append({
            'Minsup': minsup, 'Minconf': minconf,
            'Apriori - Số tập PB': len(ap['fi']), 'FP-Growth - Số tập PB': len(fp['fi']),
            'Apriori - Số luật': len(ap['rules']), 'FP-Growth - Số luật': len(fp['rules']),
            'Cùng tập phổ biến?': 'Có' if same_fi else 'KHÔNG',
            'Cùng bộ luật?': 'Có' if same_rules else 'KHÔNG',
            'Apriori - TG mining (s)': round(ap['t_mining'], 4),
            'FP-Growth - TG mining (s)': round(fp['t_mining'], 4),
            'Apriori - TG tổng (s)': round(ap['t_total'], 4),
            'FP-Growth - TG tổng (s)': round(fp['t_total'], 4),
            'Speedup (Apriori/FP)': round(speedup, 2),
            'Nhanh hơn': 'FP-Growth' if fp['t_total'] < ap['t_total'] else 'Apriori'})
        print(f'minsup={minsup} minconf={minconf} | FI {len(ap["fi"])}/{len(fp["fi"])} | '
              f'Rules {len(ap["rules"])}/{len(fp["rules"])} | same_fi={same_fi} same_rules={same_rules} | '
              f'{ap["t_total"]:.3f}s vs {fp["t_total"]:.3f}s')


# %% ===================== 3. BA BẢNG =====================
df_apriori  = pd.DataFrame(rows_ap)
df_fpgrowth = pd.DataFrame(rows_fp)
df_compare  = pd.DataFrame(rows_cmp)

print('\n=== BẢNG 1: APRIORI ===');   display(df_apriori)
print('\n=== BẢNG 2: FP-GROWTH ===');  display(df_fpgrowth)
print('\n=== BẢNG 3: SO SÁNH TRỰC TIẾP ==='); display(df_compare)

all_same = (df_compare['Cùng tập phổ biến?'].eq('Có').all()
            and df_compare['Cùng bộ luật?'].eq('Có').all())
n_fp = (df_compare['Nhanh hơn'] == 'FP-Growth').sum()
print('\n=== KẾT LUẬN TỰ ĐỘNG ===')
print('- Kết quả (tập phổ biến + luật) giống nhau ở cả 9 cấu hình:', 'CÓ' if all_same else 'KHÔNG – cần kiểm tra lại')
print(f'- FP-Growth nhanh hơn ở {n_fp}/9 cấu hình; speedup TB = {df_compare["Speedup (Apriori/FP)"].mean():.2f}x '
      f'(min {df_compare["Speedup (Apriori/FP)"].min():.2f}x, max {df_compare["Speedup (Apriori/FP)"].max():.2f}x)')
print('  (speedup < 1 nghĩa là Apriori nhanh hơn; chỉ kết luận cho tập dữ liệu + thư viện này)')


# %% ===================== 4. BIỂU ĐỒ SO SÁNH =====================
def ve_bieu_do(dfc, out_png='so_sanh_Apriori_FPGrowth.png'):
    g = dfc.groupby('Minsup').mean(numeric_only=True)
    x = np.arange(len(g)); w = 0.36
    labels = [str(m) for m in g.index]
    C_AP, C_FP = '#2E75B6', '#ED7D31'

    fig, ax = plt.subplots(2, 2, figsize=(14, 9))

    def cap_bar(a, col_ap, col_fp, title, ylab):
        b1 = a.bar(x - w/2, g[col_ap], w, label='Apriori', color=C_AP)
        b2 = a.bar(x + w/2, g[col_fp], w, label='FP-Growth', color=C_FP)
        a.bar_label(b1, fmt='%.3f', fontsize=8); a.bar_label(b2, fmt='%.3f', fontsize=8)
        a.set_title(title); a.set_ylabel(ylab)
        a.set_xticks(x); a.set_xticklabels(labels)

    # (a) thời gian tổng
    cap_bar(ax[0, 0], 'Apriori - TG tổng (s)', 'FP-Growth - TG tổng (s)',
            'Thời gian tổng (mining + sinh luật) theo minsup', 'Giây (TB 3 mức minconf)')
    # (b) thời gian mining
    cap_bar(ax[0, 1], 'Apriori - TG mining (s)', 'FP-Growth - TG mining (s)',
            'Thời gian mining thuần theo minsup', 'Giây (TB 3 mức minconf)')

    # (c) số luật theo minsup, mỗi minconf một đường (2 thuật toán trùng nhau)
    for mc, d in dfc.groupby('Minconf'):
        ax[1, 0].plot(d['Minsup'], d['Apriori - Số luật'], marker='o', label=f'minconf = {mc}')
    ax[1, 0].set_yscale('log'); ax[1, 0].set_xticks(sorted(dfc['Minsup'].unique()))
    ax[1, 0].set_title('Số luật theo minsup (Apriori = FP-Growth)')
    ax[1, 0].set_ylabel('Số luật (trục log)')

    # (d) số tập phổ biến theo minsup
    fi_n = dfc.groupby('Minsup')['Apriori - Số tập PB'].first()
    b = ax[1, 1].bar([str(m) for m in fi_n.index], fi_n.values, color='#70AD47')
    ax[1, 1].bar_label(b, fmt='%d')
    ax[1, 1].set_title('Số tập phổ biến theo minsup (Apriori = FP-Growth)')
    ax[1, 1].set_ylabel('Số tập phổ biến')

    for a in ax.ravel():
        a.set_xlabel('Minsup'); a.grid(alpha=0.3); a.legend(fontsize=8) if a.get_legend_handles_labels()[0] else None
    fig.suptitle('So sánh Apriori vs FP-Growth – tập Mushroom', fontsize=14, fontweight='bold')
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    fig.savefig(out_png, dpi=150)
    plt.show()
    print(f'=> Đã lưu biểu đồ: {out_png}')

ve_bieu_do(df_compare)


# %% ===================== 5. LỌC LUẬT & CHỌN LUẬT TIÊU BIỂU =====================
fi = fpgrowth(df_encoded, min_support=SEL_MINSUP, use_colnames=True)
rules = association_rules(fi, metric='confidence', min_threshold=SEL_MINCONF)
n0 = len(rules)

# B1. Bỏ luật lift ≈ 1 (vế trái và vế phải gần như độc lập)
rules = rules[(rules['lift'] < LIFT_LOW) | (rules['lift'] > LIFT_HIGH)].copy()
n1 = len(rules)

# B2. Bỏ luật dư thừa: X -> Y bị loại nếu có luật tổng quát hơn X' -> Y (X' ⊂ X)
#     với confidence >= confidence của luật dài hơn
rules['k'] = rules['antecedents'].apply(len)
rules = rules.sort_values('k')
best, keep = {}, []
for idx, r in rules.iterrows():
    ante, cons, conf = r['antecedents'], r['consequents'], r['confidence']
    redundant = any(a < ante and c >= conf - 1e-12 for a, c in best.get(cons, []))
    if not redundant:
        keep.append(idx)
        best.setdefault(cons, []).append((ante, conf))
rules = rules.loc[keep].copy()
n2 = len(rules)

# B3. Bỏ luật "đảo chiều" trùng tập item: ưu tiên chiều có class ở vế phải, rồi confidence cao
rules['itemset'] = rules.apply(lambda r: frozenset(r['antecedents'] | r['consequents']), axis=1)
rules['cls_rhs'] = rules['consequents'].apply(lambda s: any(i.startswith('class_') for i in s))
rules = (rules.sort_values(['cls_rhs', 'confidence', 'lift'], ascending=False)
              .drop_duplicates('itemset', keep='first'))
n3 = len(rules)
print(f'Số luật: ban đầu {n0} -> bỏ lift≈1: {n1} -> bỏ dư thừa: {n2} -> bỏ trùng tập: {n3}')

# B4. Chọn luật tiêu biểu
def fmt(s): return ' & '.join(sorted(s))
rules['score'] = rules['confidence'] * rules['lift']
only_one = rules['consequents'].apply(len) == 1
pick = []
for cls in ['class_e', 'class_p']:
    m = only_one & rules['consequents'].apply(lambda s, c=cls: c in s)
    pick.append(rules[m].sort_values(['score', 'support'], ascending=False).head(N_CLASS_PER_SIDE))
oth = rules[~rules['cls_rhs'] & only_one].sort_values('score', ascending=False)
oth = oth.assign(attr=oth['consequents'].apply(lambda s: next(iter(s)).split('_')[0]))
pick.append(oth.drop_duplicates('attr').head(N_OTHER))      # mỗi thuộc tính vế phải 1 luật -> đa dạng

sel = pd.concat(pick)
df_select = pd.DataFrame({
    'Luật (A -> C)': sel['antecedents'].apply(fmt) + '  ->  ' + sel['consequents'].apply(fmt),
    'Support': sel['support'].round(4), 'Confidence': sel['confidence'].round(4),
    'Lift': sel['lift'].round(3),
    'Nhóm': ['Edible' if 'class_e' in c else 'Poisonous' if 'class_p' in c else 'Khác'
             for c in sel['consequents']]}).reset_index(drop=True)
df_select.index += 1
print(f'\n=== {len(df_select)} LUẬT TIÊU BIỂU (gửi Kiệt diễn giải) ===')
display(df_select)


# %% ===================== 6. XUẤT EXCEL =====================
OUT = 'Tuan2_Apriori_vs_FPGrowth_Ketqua.xlsx'
with pd.ExcelWriter(OUT, engine='openpyxl') as w:
    df_apriori.to_excel(w, sheet_name='1. Apriori', index=False)
    df_fpgrowth.to_excel(w, sheet_name='2. FP-Growth', index=False)
    df_compare.to_excel(w, sheet_name='3. So sánh trực tiếp', index=False)
    df_select.to_excel(w, sheet_name='4. Luật tiêu biểu', index_label='STT')
    pd.DataFrame({'Bước lọc': ['Ban đầu', 'Sau bỏ lift≈1', 'Sau bỏ dư thừa', 'Sau bỏ trùng tập'],
                  'Số luật': [n0, n1, n2, n3]}).to_excel(w, sheet_name='5. Thống kê lọc', index=False)
    from openpyxl.drawing.image import Image as XLImage
    ws = w.book.create_sheet('6. Biểu đồ')
    ws.add_image(XLImage('so_sanh_Apriori_FPGrowth.png'), 'A1')
df_compare.to_csv('FPGrowth_vs_Apriori_Comparison.csv', index=False, encoding='utf-8-sig')
print(f"=> Đã xuất '{OUT}' và 'FPGrowth_vs_Apriori_Comparison.csv'")
```
